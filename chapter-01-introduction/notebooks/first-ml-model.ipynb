````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

print("Ready to learn!")
````

## 1. Prepare examples

Each **email** is an input. Each **label** is the answer we want the model to learn from: `spam` or `not spam`. We use short synthetic examples so you can see everything on one screen.

````python
train_emails = [
    "claim your free gift today",
    "win a prize now",
    "urgent claim your reward",
    "free cash prize waiting",
    "congratulations you have won a gift",
    "get your bonus reward now",
    "exclusive free offer for you",
    "click here to claim a prize",
    "can we meet after class",
    "please review the class notes",
    "the meeting starts at noon",
    "i sent the homework yesterday",
    "thank you for your help",
    "let us discuss the project",
    "see you at the library",
    "please send the report tomorrow",
]
train_labels = ["spam"] * 8 + ["not spam"] * 8
print("Training examples:", len(train_emails))
print("Example:", train_emails[0], "→", train_labels[0])
````

## 2. Train the model

The computer needs numbers, so `CountVectorizer` turns each email into word-count features. `MultinomialNB` learns which word patterns are associated with each label.

Run the cell once. You have now performed **supervised learning**.

````python
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(train_emails)
model = MultinomialNB()
model.fit(X_train, train_labels)
print("Training complete!")
````

## 3. Test on NEW examples

These messages were not part of the training set. This is much closer to how a model is actually used. (The test set is deliberately tiny, so don't take the accuracy as a reliable general-performance estimate.)

````python
test_emails = [
    "claim your free reward",
    "can you send the homework",
    "a bonus prize for you",
    "please attend the project meeting",
    "i appreciate your help today",
    "win your gift now",
]
test_labels = ["spam", "not spam", "spam", "not spam", "not spam", "spam"]
predictions = model.predict(vectorizer.transform(test_emails))
for email, expected, predicted in zip(test_emails, test_labels, predictions):
    print(f"{email:39} expected={expected:9} predicted={predicted}")
print(f"\nCorrect on this tiny test set: {accuracy_score(test_labels, predictions):.0%}")
````

## 4. Investigate surprising messages

Now try messages that look different from our training examples. A legitimate message may include `free`, and an unwanted message may avoid familiar spam words.

**These messages have no correct labels provided here:** your job is to decide what a safe email app should do and compare that judgment with the model output.

````python
unusual_emails = [
    "are you free to meet tomorrow",  # likely legitimate
    "my new address is listed below",  # likely legitimate
    "verify the account immediately to avoid suspension",  # potentially phishing
    "you have been selected for an exclusive meeting",  # ambiguous
]
for email, prediction in zip(unusual_emails, model.predict(vectorizer.transform(unusual_emails))):
    print(f"{prediction:9} ← {email}")
````

## 5. Try your own email

Change `my_email`, re-run the cell, and compare the answer to what you expect.

**Think:** Does the model have enough examples to handle scams in other languages, messages that contain attachments, or new attacks? No—this tiny example does not.

````python
my_email = "can we meet about the free workshop"
my_prediction = model.predict(vectorizer.transform([my_email]))[0]
print("Model predicts:", my_prediction)
````
