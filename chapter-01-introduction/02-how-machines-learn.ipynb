## 1. Four ingredients of training

Think of training as repeated practice with feedback:

1. **Data:** example emails and their correct categories.
2. **Model:** a system with adjustable numerical parameters.
3. **Loss function:** a measure of how poorly the model's predictions match the desired targets.
4. **Optimizer:** a method for adjusting the parameters to try to reduce loss.

A common optimizer family uses **gradient descent**, which you'll study mathematically in Chapter 2.

![Training cycles through data, predictions, loss, and parameter updates.](images/training-loop.svg)

### One email, step by step

- **Input:** "You have won a free gift."
- **Correct label:** Spam.
- **Initial prediction:** Not spam.
- **Loss:** the model's prediction does not match the desired target.
- **Update:** the optimizer changes model parameters using feedback from one or more training examples.
- **Repeat:** after many updates, we hope the model performs better on new emails.

An update is not a magic correction: it can improve overall performance while making an individual example worse.

![The difference between a training label and prediction contributes to loss.](images/loss-feedback.svg)

## 2. How does optimization know what to change?

Imagine standing on a hillside and trying to find a lower point. **Gradient descent** uses information about how the loss changes with parameters to decide which direction to adjust them. Real models have many parameters; the graph below is only an intuition.

![Points descend along an illustrative loss curve toward a lower-loss setting.](images/gradient-descent-intuition.svg)

### Optional mini-experiment: does a lower loss look better?

This is a **toy mathematical example**, not a spam classifier. The parameter `weight` is adjusted to reduce `(weight − 2)²`. We will learn derivatives and gradient descent properly in Chapter 2.

```python
weight = 0.0
learning_rate = 0.2
for step in range(6):
    loss = (weight - 2) ** 2
    print(f"Step {step}: weight={weight:.3f}, loss={loss:.3f}")
    gradient = 2 * (weight - 2)  # Slope of this toy loss
    weight -= learning_rate * gradient
```

## 3. Three useful ways machines learn

Machine learning isn't just one kind of training. The differences mainly concern **where the learning signal comes from**.

![Supervised, self-supervised, and reinforcement learning compared.](images/three-learning-methods.svg)

### A. Supervised learning: labeled examples

Our spam detector uses labeled messages: the model sees the input, and training uses its known category. Other examples include categorizing images and predicting prices.

### B. Self-supervised learning: learn targets from the data

Take: **"The student opened the ..."** A language model can learn to predict the next **token** (a word or part of a word) using the following text as the training target. People do not have to label every sentence by hand. Many language models use this type of learning during pretraining.

**Important:** skill at predicting text does not guarantee truthfulness or good judgement.

### C. Reinforcement learning: actions and rewards

A robot navigating a maze can choose actions and receive rewards from its environment. Through repeated interactions, it can learn a **policy**: a strategy for choosing actions. Reward-based methods are also relevant to training and fine-tuning some AI assistants.

**One more term:** unsupervised learning finds patterns in data without a target label per example—for instance, grouping similar articles.

## 4. What makes deep learning "deep"?

**Deep learning (DL)** is a subset of machine learning that uses neural networks with multiple layers of learned transformations. Each layer receives numerical information and passes a transformed representation to another layer.

![A conceptual neural network processes inputs through multiple layers to predictions.](images/neural-network-layers.svg)

For images, an early layer might learn simple shapes and a later layer might combine information into features useful for classification. This is a **helpful intuition**, not a fixed rule for every neural network.

Deep learning often learns representations from data, reducing the need to manually invent every feature.

![An illustrative progression from image pixels to learned representations and output.](images/representation-learning.svg)

## 5. Where does AI fit?

**Artificial intelligence (AI)** is the broader field of building systems that perform tasks associated with intelligence. **Machine learning** is one approach; **deep learning** is one family of ML methods. A modern **Transformer** is a deep learning architecture widely used in large language models.

![Illustrative hierarchy of artificial intelligence, machine learning, deep learning, and modern language models.](images/ai-ml-dl-hierarchy.svg)

You do **not** need to understand attention heads, embeddings or matrix multiplication yet. Those are later topics. For now, keep the core distinction: **models learn adjustable patterns; deep neural networks learn layered representations.**
