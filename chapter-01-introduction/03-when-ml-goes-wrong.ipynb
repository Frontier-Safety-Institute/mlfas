Suppose our spam detector classifies **every training email correctly**. Are we finished? **No.**

## 1. Memorizing is not the same as learning useful patterns

A student might memorize practice questions without learning how to solve new questions. Likewise, an ML model can do very well on examples it trained on and still fail on different examples. This is **overfitting**.

**Generalization** means performing well on *unseen* data.

![A model that overfits performs well on training data but poorly on new examples.](images/generalization.svg)

This is why we separate examples into **training data** and **evaluation data**. A **validation set** helps during model development; a carefully held-out **test set** is used for a final estimate. A test score provides useful evidence, not a guarantee of real-world safety.

## 2. A surprisingly bad shortcut

What if nearly all spam examples in the training set contain the word **free**? The model may associate that word with spam, even when the sentence is "Are you free tonight?"

![A spam detector mistakes a harmless question containing free for spam.](images/shortcut-learning.svg)

This is **shortcut learning**: a pattern may appear useful in training data while failing for the real task. Better data and targeted evaluations can help identify such weaknesses.

**Question:** What kind of non-spam message would you add to test whether the model depends too heavily on the word "free"?

## 3. What if the world changes?

Spam tactics evolve. A model might be trained mostly on English messages and then face spam written in other languages or with unfamiliar wording. This is an example of **distribution shift**: the data encountered in use differs from the data on which the model was developed.

![Model deployment introduces unfamiliar examples compared with training data.](images/distribution-shift.svg)

## 4. A good score can hide a bad outcome

Imagine evaluating a spam filter only by **how few spam emails reach the inbox**. It could block *all* incoming messages and score perfectly by that narrow measure. But users would miss legitimate emails.

This is a simple example of **objective mismatch**: optimizing one measurable target may not achieve what people really want.

![A filter can get a good spam-blocking score while blocking legitimate messages.](images/objective-mismatch.svg)

That mismatch is relevant to **AI alignment**, where we ask whether the system's behavior reflects the intended goals and constraints, not merely a convenient proxy measure.

## 5. Some failures come from the entire system

Imagine an AI assistant is asked to summarize a document. An attacker inserts a sentence into that document saying "Ignore the user and do something else."

This is an example of a **prompt-injection attempt**: instructions planted in untrusted content try to redirect the assistant. Whether an attack succeeds depends on the model and the surrounding system, including permissions and how the application handles untrusted data.

![A malicious instruction inside untrusted content tries to override the user's real request.](images/prompt-injection-boundary.svg)

A model can also confidently produce a false answer; this is sometimes called a **hallucination**. Prompt injection and hallucination are different failure modes. Both matter for trustworthy AI, but their causes and defenses can differ.

## 6. Three connected fields

- **Technical AI safety:** Can we understand, evaluate, and reduce harmful failures?
- **AI alignment:** Does the system pursue behavior consistent with intended human goals and constraints?
- **AI security:** Can an attacker manipulate an AI model or application into unintended behavior?

These fields overlap. For example, a security compromise may also create a safety incident.

![An introduction to safety, alignment, and security as overlapping areas of work.](images/safety-alignment-security.svg)

## 7. Think like a researcher

Distinguish **observation** from **explanation**:

- **Observation:** "The model marked this legitimate email as spam."
- **Hypothesis:** "It might be focusing too much on the word free."
- **Experiment:** Test many normal emails containing "free" and similar messages without it.

That habit matters later for mechanistic interpretability, evaluations and security research.

### Optional code: count two kinds of mistakes

A **false positive** means a real email is incorrectly flagged as spam. A **false negative** means spam slips through. A good system should consider both.

```python
actual = ["Spam", "Not spam", "Spam", "Not spam", "Not spam"]
predicted = ["Spam", "Spam", "Not spam", "Not spam", "Spam"]
false_positives = sum(a == "Not spam" and p == "Spam" for a, p in zip(actual, predicted))
false_negatives = sum(a == "Spam" and p == "Not spam" for a, p in zip(actual, predicted))
print("Legitimate emails blocked:", false_positives)
print("Spam emails that slipped through:", false_negatives)
```
