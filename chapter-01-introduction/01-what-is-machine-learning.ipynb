We will follow **one running example** throughout this chapter: building a simple spam-email detector.

## 1. A problem anyone can recognize

Imagine building an email app that needs to decide which messages are **spam**. You might write a rule: *"If an email contains WIN A FREE PRIZE, mark it as spam."*

But then spammers write *"Your gift is ready"*, and a legitimate friend writes *"Did you win the prize?"*

A rigid rule may miss unwanted messages and wrongly block genuine ones. Instead of coding every condition by hand, **machine learning (ML)** adjusts a model using examples.

![Traditional rules are written by people; machine-learning parameters are learned from examples.](images/ml-vs-rules.svg)

**Definition:** Machine learning uses data and learning algorithms to adjust a model so it can make predictions on new inputs. People still design the task, select data, and decide how to evaluate the system.

**Think about it:** Would you prefer hand-written rules for adding two numbers or learning examples of addition? Ordinary programming is better when the rule is already exact and reliable.

## 2. Data: what does the computer see?

For this spam detector, each training example has two main parts:

- **Input:** the email text that the model sees.
- **Label:** the correct category supplied for training, such as *Spam* or *Not spam*.

A collection of examples is called a **dataset**.

![Four email examples shown with their training labels.](images/labeled-dataset.svg)

The computer does not understand text the way a person does. It must process a **numerical representation** of the input. A **feature** is a useful property of an example. A very simple email model might ask whether certain words appear; larger models learn richer representations.

![Email text becomes numeric features such as whether prize or meeting appears.](images/text-to-features.svg)

### Try this (optional)

```python
emails = [
    {"text": "Claim your prize now", "label": "Spam"},
    {"text": "Please send the class notes", "label": "Not spam"},
]
for email in emails:
    print(f"Input: {email['text']}  →  Label: {email['label']}")
```

## 3. What is a model?

A **model** is the component that takes an input and produces a **prediction**. It contains adjustable numbers called **parameters** that influence what it predicts. During training, a learning algorithm adjusts these parameters.

![A trained spam detector predicts that a previously unseen email is spam.](images/model-prediction.svg)

The point is **not** to memorize every training email. We want the model to identify patterns that remain useful on *new* emails. Whether it succeeds is something we have to test.

## 4. Training versus inference

These are different activities:

- **Training:** learn from examples, compare predictions with answers, and update model parameters.
- **Inference:** use the trained model to make a prediction for a new email, normally without changing the model parameters.

![Training changes model parameters, while inference ordinarily uses fixed trained parameters.](images/training-vs-inference.svg)

This distinction is important for later AI safety lessons: **an AI assistant producing a response is not necessarily learning new weights from that conversation**. A system may separately store memories or retrieve information; neither automatically means the underlying model is being retrained.

## 5. What can models predict?

Two common tasks are **classification** (choose a category) and **regression** (predict a number). Both can use machine learning, but their outputs are different.

![Classification predicts categories while regression predicts numerical values.](images/classification-vs-regression.svg)
